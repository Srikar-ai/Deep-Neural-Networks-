```python
import numpy as np 
import pandas as pd 
import torch 
import torch.nn as nn
```

```python
class MLP:


    def __init__(self,architecture,X,y,lr,epochs,use_case):


        self.layers = []
        for i in range(len(architecture) - 2):
            self.layers.append( nn.Linear( architecture[i], architecture[i + 1] ) ) 
            self.layers.append(nn.ReLU())

        self.layers.append( nn.Linear( architecture[-2], architecture[-1] ) )
        self.net = nn.Sequential(*self.layers)
        self.X = X 
        self.y = y 
        self.lr = lr
        self.epochs = epochs
        self.use_case = use_case 
        self.loss = None
        self.optimizer = torch.optim.SGD(
                    self.net.parameters(),
                    lr = self.lr 
                )
        if self.use_case == "Regression":
            self.loss = nn.MSELoss()
        if self.use_case == "Classification":
            self.loss = nn.CrossEntropyLoss()

    def forward(self):
        return self.net(self.X)

    def backward(self,y_pred):

        loss = self.loss(y_pred,self.y)
        self.optimizer.zero_grad()
        loss.backward()
        self.optimizer.step()

        return loss 

    def fit(self):

        for i in range(self.epochs):

            y_pred = self.forward()
            loss = self.backward(y_pred)
            print(loss)
```

```python
import torch
import numpy as np

np.random.seed(42)

# 100 samples, 2 features
X = np.random.randn(100, 2)

# True relationship
y = 3 * X[:, 0] + 2 * X[:, 1] + 5

# Add small noise
y += np.random.randn(100) * 0.2

# Make y shape (100, 1)
y = y.reshape(-1, 1)

X = torch.tensor(X, dtype=torch.float32)
y = torch.tensor(y, dtype=torch.float32)

print("X shape:", X.shape)
print("y shape:", y.shape)
```

```python
model = MLP(
    architecture=[2, 8, 4, 1],
    X=X,
    y=y,
    epochs=5000,
    lr=0.01,
    use_case="Regression"
)
model.fit()

y_pred = model.forward()

print("Actual:")
print(y[:10].flatten())

print("\nPredicted:")
print(y_pred[:10].flatten())

rmse = torch.sqrt(torch.mean((y_pred - y) ** 2))

print("RMSE:", rmse.item())
```

```python
import numpy as np
import torch

np.random.seed(42)

n = 100

# Class 0
X0 = np.random.randn(n, 2) + np.array([-2, -2])
y0 = np.zeros(n, dtype=int)

# Class 1
X1 = np.random.randn(n, 2) + np.array([2, -2])
y1 = np.ones(n, dtype=int)

# Class 2
X2 = np.random.randn(n, 2) + np.array([0, 2])
y2 = np.full(n, 2, dtype=int)

# Combine
X = np.vstack([X0, X1, X2])
y = np.concatenate([y0, y1, y2])

# Convert NumPy → PyTorch
X = torch.tensor(X, dtype=torch.float32)
y = torch.tensor(y, dtype=torch.long)

print("X shape:", X.shape)
print("y shape:", y.shape)
print("X dtype:", X.dtype)
print("y dtype:", y.dtype)
```

```python
model = MLP(
    architecture=[2, 8, 8, 3],
    X=X,
    y=y,
    epochs=5000,
    lr=0.01,
    use_case="Classification"
)

model.fit()

# Forward pass gives logits
logits = model.forward()

# Convert logits → predicted class
predictions = torch.argmax(logits, dim=1)

# Accuracy
accuracy = (predictions == y).float().mean()

print("Accuracy:", accuracy.item())
```

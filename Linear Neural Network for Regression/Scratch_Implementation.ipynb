```python
import numpy as np 
import pandas as pd 
from time import time 
```

```python
class LNN_Regression:

    def __init__(self,X,y,epochs,lr,reg_constant = None,reg_type = None):
        self.X = np.array(X)
        self.W = np.random.rand(X.shape[1])
        self.b = np.zeros(1)
        self.y = np.array(y) 
        self.epochs = epochs
        self.lr = lr 
        self.reg_constant = reg_constant
        self.type = reg_type


    def forward(self):
        return self.X @ self.W + self.b

    def fit(self):
        for _ in range(self.epochs):
            y_pred = self.forward()
            loss = self.loss(self.y,y_pred)
            print(f"Loss @ Epoch {_} : ",loss)
            error = y_pred  - self.y    
            # print(error.shape)
            # print(self.X.T.shape)
            dL_dw = (self.X.T @ error) * (2/self.X.shape[0])
            if self.type == "l1":
                dL_dw += (self.reg_constant) * np.sign(self.W)
            if self.type == "l2":
                dL_dw += (self.reg_constant) * self.W 

            self.W = self.W - self.lr * dL_dw
            self.b = self.b - self.lr * np.sum(error) * (2/self.X.shape[0])

    def loss(self,y,y_pred):
        penalty = 0 
        if self.type == "l1":
            penalty = self.L1_loss()
        if self.type == "l2":
            penalty = self.L2_loss() 
        return np.mean((y-y_pred)**2) + penalty

    def L2_loss(self):
        return self.reg_constant * (0.5) * np.sum((self.W)**2)
    def L1_loss(self):
        return self.reg_constant * np.sum(abs(self.W))
```

### Linear  Regressino without Regularization

```python

start = time()
X = np.arange(1, 11).reshape(-1, 1)
y = np.array([3, 4, 6, 7, 13, 18, 20, 23, 21, 25])

model = LNN_Regression(X, y, epochs=5000, lr=0.01)
model.fit()

end = time()

print("Total time : ", end-start)
print("Weights:", model.W)
print("Bias:", model.b)
print("Predictions:", model.forward())
```

### Linear  Regressino with L1 Regualarization

```python

start = time()
X = np.arange(1, 11).reshape(-1, 1)
y = np.array([3, 4, 6, 7, 13, 18, 20, 23, 21, 25])

model = LNN_Regression(X, y, epochs=5000, lr=0.01,reg_constant=0.01,reg_type="l1")
model.fit()

end = time()

print("Total time : ", end-start)
print("Weights:", model.W)
print("Bias:", model.b)
print("Predictions:", model.forward())
```

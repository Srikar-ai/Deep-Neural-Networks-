```python
import numpy as np 
from collections import Counter
```

### Unigram Model 

```python
class UnigramModel:


    def __init__(self):

        self.word_counts = Counter()
        self.total_words = 0 
        self.vocab = set()
        self.probabilities = {}

    def train(self,X):

        tokens = X.split(" ")

        self.word_counts = Counter(tokens)
        self.total_words = len(tokens)

        for i in tokens:
            self.vocab.add(i)

        for i in tokens:
            self.probabilities[i] = self.word_counts[i]/(self.total_words)

    def predict_next(self, text=None):
        if not self.probabilities:
            return None

        return max(
            self.probabilities,
            key=self.probabilities.get
        )
            
```

### Model Evaluation 

```python
model = UnigramModel()
model.train("the cat sat the cat slept the dog ran")

print(model.word_counts)
print(model.total_words)
print(model.vocab)
print(model.probabilities)

print("Predicted Next Token : ",model.predict_next("the cat"))
```
